```python
import pandas as pd
import json

# Definimos la carpeta donde están los datos
RUTA_DATOS = "Archivos GO/"

# 1. Cargamos todos los archivos CSV en DataFrames de pandas
df_calidad_atrib = pd.read_csv(RUTA_DATOS + "calidad_atributos.csv")
df_calidad_llenado = pd.read_csv(RUTA_DATOS + "calidad_llenado.csv")
df_capacidad = pd.read_csv(RUTA_DATOS + "capacidad_horizonte.csv")
df_compromisos = pd.read_csv(RUTA_DATOS + "compromisos_semana_focal.csv")
df_demanda = pd.read_csv(RUTA_DATOS + "demanda_semanal.csv")
df_especificaciones = pd.read_csv(RUTA_DATOS + "especificaciones_calidad.csv")
df_inventario = pd.read_csv(RUTA_DATOS + "inventario_inicial.csv")
df_maestro_sku = pd.read_csv(RUTA_DATOS + "maestro_sku.csv")
df_setups = pd.read_csv(RUTA_DATOS + "setups_cip.csv")

# 2. Cargamos también el archivo JSON como un diccionario de Python
with open(RUTA_DATOS + "metadatos.json", "r", encoding="utf-8") as archivo_json:
    metadatos = json.load(archivo_json)

# 3. Verificamos que todo haya cargado bien mostrando una vista previa
print("--- Metadatos del proyecto ---")
print(metadatos)

print("\n--- Primeras filas de Maestro SKU ---")
display(df_maestro_sku.head())

print("\n--- Primeras filas de Demanda Semanal ---")
display(df_demanda.head())
```
